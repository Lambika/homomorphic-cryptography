```python
from pydub import AudioSegment
from pydub.silence import detect_silence

audio = AudioSegment.from_file('python_banane_ia_2sdelay.mp3')  # adapte le chemin et le nom

silence_thresh = audio.dBFS - 16


silences = detect_silence(audio, min_silence_len=500, silence_thresh=silence_thresh)

silence_durations = [(end - start) / 1000 for start, end in silences]
total_silence_duration = sum(silence_durations)

print(f"Nombre de silences détectés : {len(silences)}")
print(f"Durée totale du silence : {total_silence_duration:.2f} secondes")
```
