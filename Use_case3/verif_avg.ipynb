```python
def load_model_file(filepath):
    model = {}
    with open(filepath, "r") as f:
        lines = [line.strip() for line in f if line.strip()]
        for i in range(0, len(lines), 2):
            name = lines[i]
            values = list(map(float, lines[i + 1].split()))
            model[name] = values
    return model

def average_models(model1, model2):
    avg = {}
    for key in model1:
        v1 = model1[key]
        v2 = model2[key]
        avg[key] = [(a + b) / 2 for a, b in zip(v1, v2)]
    return avg

def print_model(model):
    for key, values in model.items():
        print(key)
        print(" ".join(f"{v:.6f}" for v in values))
        print()

# === Usage ===
file1 = "params_client1.txt"
file2 = "params_client2.txt"

model1 = load_model_file(file1)
model2 = load_model_file(file2)
avg_model = average_models(model1, model2)

print("=== Moyenne des deux modèles ===")
print_model(avg_model)
```
